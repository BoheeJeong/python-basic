### 3.웹 데이터 요청하기
명세서의 요구에 따라 헤더에  client-id 와 secret 항목을 추가한다 (위에서 변수 생성 완료)
그리고 post 형식으로 웹 서버와 통신한다

*POST 전송 - 요청 데이터를 JSON 으로 전달하기 위해서 딕셔너리 데이터를 문자열로 변환한다


<GET VS POST>
- GET은 데이터를 “주소(URL)”에만 붙여서 보내고, URL 에 모든 정보가 노출되지만, POST 는 
POST는 데이터를 “요청 본문(body)”에 JSON이나 form 형식으로 담아서 보내기 떄문에 
URL은 짧고 깔끔하고, 실제 내용(JSON)은 “본문(body)”에 들어있음.
즉, URL에는 아무것도 안 보이지만, 서버는 body를 읽어서 처리하며 대용량, 복잡한 데이터도 전달 가능






```python
with requests.Session() as session:
    session.headers.update({
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36",
        "X-Naver-Client-Id": clientId,
        "X-Naver-Client-Secret": clientSecret
    })

    r = session.post(url, data=json.dumps(data))  #url 주소로 post 요청을 보내는데, data 라는 내용을 json 형식으로 본문에 담아 전송하고 그 결과를 r 이라는 변수에 저장한다

    if r.status_code != 200:
        msg = "[%d Error] %s 에러가 발생함" % (r.status_code, r.reason)
        raise Exception(msg)

print(r)
```

### 4.응답 결과 확인
1차적으로 가져와진 데이터 형태는 리스트로 딕셔너리가 감싸져 있고, 또 그 안에 data 키는 또 다른 리스트와 딕셔너리로 감싸져 있음. 우리가 필요한 ratio 데이터를 가져오기 위해서는 2차원 데이터를 1차원으로 전처리를 해줘야함

### 5.응답결과 재구성
*2차원 구조를 1차원 구조로 전처리

바로 위의 결과가  mydict["results"] 였으니, 우선 1차적으로 해당 항목의 리스트를 풀어주고 나면
우리가 필요한 데이터인 ratio 에 접근하기 위해서 data key 에 enumerate 로 접근한다

그러면 위에서 title 이란 변수에 각 키워드 이름을 넣어준 상태이니,
period 와 ratio 만 각 딕셔너리에서 키값으로 접근을 해주고

비워져있는 resultset 리스트에 해당 행을 하나씩 extend 해준다



```python
resultset = []

for item in mydict["results"]:
    title = item["title"]
    data = item["data"]

    for i, v in enumerate(data):
        new_data = {"title": title, "period": v["period"], "ratio": v["ratio"]}
        data[i] = new_data

    resultset.extend(data)

resultset
```
