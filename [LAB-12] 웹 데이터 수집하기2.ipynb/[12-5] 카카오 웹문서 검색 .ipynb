### 1. 필요한 라이브러리 참조하기

```python
import requests
from pandas import DataFrame
```

### 요청 정보 확인

```python
url = "https://dapi.kakao.com/v2/search/web"

query =['고슴도치','햄스터',"미어캣"]
page = 1
size=50

key ="ba2b91a53d05dab15c540eff6f1c5565"
```

### 웹 데이터 요청하기 (동기식)


```python
#데이터를 담을 비워진 리스트 생성
resultset = []


with requests.Session() as session:
  session.headers.update({
    "User-Agent":"Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/51.0.2704.103 Safari/537.36",
    "Authorization" : "KakaoAK %s" %key
  })



# 키워드 리스트 탐색
for i,query_item in enumerate(query) :
  #각 키워드마다 쪽수 페이지 1-20 반복
  for v in range(1,21):
    r=session.get(url,params={"query":query_item,"page":v,"size":size})    
    

    if r.status_code ==200:
      print("%s 페이지 %s 데이터가 수신 완료되었습니다" %(v,query_item))  
    

    #HTTP 상태값이 200이 아닌 경우는 강제로 에러를 발생시켜 코드의 진행을 중단시킴
    if r.status_code !=200:
      msg ="[%d Error] %s 에러가 발생함" % (r.status_code,r.reason)
      print(msg)
      continue
```

```python
r.encoding = "utf-8"
mydict = r.json()
print(mydict)
```
