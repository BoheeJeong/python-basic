### STEP 1. 라이브러리 참조하기

```python
import requests
from pandas import DataFrame
import json
```

### STEP 2. 요청 정보 확인

```python
#인증키
KEY = "6b6c4e4849626f683933564c44446a"
# 필수 파라미터
TYPE='json'
SERVICE= "StationAdresTelno"
START_INDEX = 1
END_INDEX=1000


#요청 URL path 형식
base_url = "http://openapi.seoul.go.kr:8088"
path = f"/{KEY}/{TYPE}/{SERVICE}/{START_INDEX}/{END_INDEX}/"
url = base_url+path

print(url)
```

### 3. 웹 데이터 요청하기

```python
with requests.Session() as session:

  #세션 객체에 웹 브라우저 정보 (UserAgent) 주입 (웹서버가 파이썬 프로그램을 정상적인 웹 브라우저로 여기도록)
  session.headers.update({"User-Agent":"Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/51.0.2704.103 Safari/537.36"
                          })
  

  r=session.get(url)    
  print(r.url)

  if r.status_code !=200:
    msg ="[%d Error] %s 에러가 발생함" % (r.status_code,r.reason)
    raise Exception(msg)

print(r) # HTTP 통신 상태 확인

r.encoding ="utf-8"
mydict = r.json()
resultset = mydict["StationAdresTelno"]["row"]
print(mydict["StationAdresTelno"]["row"])


```

### 4. 지하철역 주소 데이터 응답 결과 구조 확인

```python
metro_df = DataFrame(resultset)
metro_df
```

```python
#type =parcel 에 해당하는 주소인 OLD_ADDR 활용, 해당 주소 리스트를 국토교통부 API에 활용하여 주소를 좌표로 변환할 예정
OLD_ADDR_list =[]

for v,item in enumerate(resultset):

  OLD_ADDR_list.append(item["OLD_ADDR"])

print(OLD_ADDR_list)

```

### [2] 국토교통부 V WORLD OPEN API 활용 | 주소를 좌표로 변환해보기

- STEP 0. 라이브러리 참조하기 (생략)
- STEP 1. 요청 정보 확인
- STEP 2. 웹 데이터 요청하기
- STEP 3. 데이터 전처리
- STEP 4. 데이터 프레임 형태로 확인하기

### STEP 1. 요청 정보 확인

```python
geo_url ="https://api.vworld.kr/req/address?"   #쿼리 스트링 파라미터 방식
geo_key = "1C1C7AC9-4A35-3854-BFE7-D91FAEDD655E"
geo_list =[]

for address_info in OLD_ADDR_list:

	params = {
		"service": "address",
		"request": "getcoord",
		"crs": "epsg:4326",
		"address": address_info,
		"format": "json",
		"type": "parcel",
		"key": geo_key,
	}
	response = requests.get(geo_url, params=params)
	geo_list.append(response.url)
	
print(geo_list) 
```

### STEP 2. 웹 데이터 요청하기

```python
geo_result_dict_list =[]

with requests.Session() as session2:
  session.headers.update({"User-Agent":"Mozilla/5.0 (X11; Linux x86_64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/51.0.2704.103 Safari/537.36"
                          })
  

  #앞에서 준비한 주소 url 리스트를 탐색하며 접속
for e in geo_list:
  r2=session.get(e)    
  
  if r2.status_code !=200:
    msg ="[%d Error] %s 에러가 발생함" % (r2.status_code,r2.reason)
    print(msg)
    continue


  r2.encoding ="utf-8"
  geo_mydict = r2.json()
  geo_result_dict_list.append(geo_mydict)

#국토교통부 전체 데이터 확인
print(geo_result_dict_list)  
```

### STEP 3. 데이터 전처리
- 전처리 0) status 가 ok 인 경우에 대해서만 처리한다
  
- 전처리 1) geo_result_dict_list 리스트를 탐색해서 안에 있는 딕셔너리에 접근

- 전처리 2) 필요한 데이터만 따로 데이터 프레임 형태로 만들어내기 > 리스트 안에 딕셔너리들이 원소로 들어있어야 함
비워져 있는 리스트(result_list)를 준비하고, {address: 주소 , x: x좌표,y:좌표} 형태의 원소들을 append 해주자


```python
result_list =[]
for y,item in enumerate(geo_result_dict_list):

  # staus 가 OK 인 경우에 대해서만 처리한다
  if item['response']['status'] =="OK":
    
    a= item['response']['input'] 
    b=item['response']['result']

    # 데이터 확인용 : print(y+1, a['address'], "의", "x의 좌표는", b['point']['x'] ,",y의 좌표는", b['point']['y'] )
    dict ={"OLD_ADDR":a['address'],"longitude":b['point']['x'],"latitude":b['point']['y'] }
    result_list.append(dict)


print(result_list)
    

```

### STEP4. 데이터 프레임으로 생성하기
주소지 + X 좌표 + Y 좌표 정보를 데이터 프레임으로 확인
이제 기존 원본 데이터에 주소지를 키값으로 하여 조인 하면 X,Y 좌표를 서울교통공사 데이터와 함께 확인할 수 있게 된다

```python
geo_df = DataFrame(result_list)
geo_df
```

### [3] metro_df 와 geo_df 를 pandas 를 활용하여 조인해 하나의 형태로 조합하기

```python
import pandas as pd

merged_df = pd.merge(geo_df, metro_df, on='OLD_ADDR', how='inner')
print(merged_df)

#subway_stations_with_coords 이름으로 저장
merged_df.to_csv("subway_stations_with_coords.csv",index=False,encoding='utf-8')
```
